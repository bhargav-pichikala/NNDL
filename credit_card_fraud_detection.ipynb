```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

data = pd.read_csv("creditcard.csv")

X = data.drop("Class", axis=1)
y = data["Class"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

model = Sequential([
    Dense(32, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(16, activation='relu'),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X_train, y_train, epochs=10, batch_size=2048, validation_split=0.2)

loss, accuracy = model.evaluate(X_test, y_test)
print("\nTest Accuracy:", accuracy)
pred_probs = model.predict(X_test)
predictions = (pred_probs > 0.5).astype(int)

print("\nDetected Fraud Transactions:\n")

fraud_found = False

for i in range(len(predictions)):

    if predictions[i] == 1:   # Show only fraud

        fraud_found = True

        actual = "Fraud" if y_test.iloc[i] == 1 else "Real"

        print(f"Transaction {i+1}: Predicted = Fraud | Actual = {actual}")

if not fraud_found:
    print("No fraudulent transactions detected.")

```
