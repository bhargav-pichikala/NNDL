```python
import numpy as np

# Hyperparameters
input_size = 3
hidden_size = 2
output_size = 1
lr = 0.2
epochs = 1000

# Initialize weights and biases
np.random.seed(1)
W1 = np.random.rand(input_size, hidden_size)
b1 = np.random.rand(hidden_size)
W2 = np.random.rand(hidden_size, output_size)
b2 = np.random.rand(output_size)

# Training data
x = np.array([[1, 8, 3]])
y = np.array([[0.9]])

for _ in range(epochs):
    # Forward pass
    z1 = x @ W1 + b1
    a1 = 1 / (1 + np.exp(-z1)) # Inlined sigmoid
    z2 = a1 @ W2 + b2
    a2 = 1 / (1 + np.exp(-z2)) # Inlined sigmoid

    # Calculate loss
    loss = (y - a2)**2

    # Backward pass
    # Inlined sigmoid_derivative: a*(1-a)
    dz2 = 2 * (a2 - y) * (a2 * (1 - a2))
    dW2 = a1.T @ dz2
    db2 = dz2.sum(axis=0)

    dz1 = (dz2 @ W2.T) * (a1 * (1 - a1))
    dW1 = x.T @ dz1
    db1 = dz1.sum(axis=0)

    # Update weights and biases
    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

print("Predicted:", a2)
print("Loss:", loss)
```

```python
import numpy as np

class FeedForwardNN:
    def __init__(self, input_size, hidden_size, output_size, lr=0.2):
        np.random.seed(1) # Added for reproducibility
        self.lr = lr
        self.W1 = np.random.rand(input_size, hidden_size)
        self.b1 = np.random.rand(hidden_size)
        self.W2 = np.random.rand(hidden_size, output_size)
        self.b2 = np.random.rand(output_size)

    def sigmoid(self, z):
        return 1/(1+np.exp(-z))

    def sigmoid_derivative(self, a):
        return a*(1-a)

    def forward(self, x):
        self.x = x
        self.z1 = x @ self.W1 + self.b1
        self.a1 = self.sigmoid(self.z1)
        self.z2 = self.a1 @ self.W2 + self.b2
        self.a2 = self.sigmoid(self.z2)
        return self.a2

    def backward(self, y):
        loss = (y - self.a2)**2

        dz2 = 2*(self.a2 - y) * self.sigmoid_derivative(self.a2)
        dW2 = self.a1.T @ dz2 # Corrected from np.outer
        db2 = dz2.sum(axis=0) # Corrected for bias sum

        dz1 = (dz2 @ self.W2.T) * self.sigmoid_derivative(self.a1)
        dW1 = self.x.T @ dz1 # Corrected from np.outer
        db1 = dz1.sum(axis=0) # Corrected for bias sum

        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1

        return loss

    def train(self, x, y, epochs):
        for _ in range(epochs):
            self.forward(x)
            loss = self.backward(y)
        return loss

x = np.array([[1, 8, 3]]) # Reshaped to 2D array
y = np.array([[0.9]])    # Reshaped to 2D array

nn = FeedForwardNN(input_size=3, hidden_size=2, output_size=1, lr=0.2) # Initialized with parameters
loss = nn.train(x, y, epochs=1000) # Added epochs

print("Predicted:", nn.a2)
print("Loss:", loss)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

model = nn.Sequential(
    nn.Linear(3, 2),
    nn.Sigmoid(),
    nn.Linear(2, 1),
    nn.Sigmoid()
)

criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.2)

x = torch.tensor([[1.0, 8.0, 3.0]])
y = torch.tensor([[0.9]])

epochs = 1000

for _ in range(epochs):
    y_pred = model(x)
    loss = criterion(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print("Predicted:", y_pred.detach().numpy())
print("Loss:", loss.item())
```
